```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataraw = pd.read_excel('Endy.xlsx')
dataraw
```

```python
#Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw['Grade'], columns='Frekuensi')
print(datafrq)
```

```python
datafrq.plot(color='yellow')
```

```python
warna = ['red', 'blue', 'green', 'orange', 'purple', 'cyan']
plt.bar(datafrq.index, datafrq['Frekuensi'], color=warna)
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.show()
```

```python
warna_terang = ['#FF0000', '#0000FF', '#00FF00', '#FF8000', '#8000FF', '#00FFFF']
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False, colors=warna_terang)
```

```python
#Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"],errors='coerce')
dt = dataraw["Final Score"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['Variance'] = dt.var()
stats['Range'] = dt.max() - dt.min()
stats['Skewness'] = dt.skew()
stats['Kurtosis'] = dt.kurtosis()
print(stats)
```
